# 1. Series 的创建

```python
# series 的创建
import pandas as pd
import numpy as np

s = pd.Series([1,2,3,4,5])
print(s)

#自定义索引
s = pd.Series([10,2,4,5,6], index=['A','B','C','D','E'])
print(s)
#定义name
s = pd.Series([10,2,4,5,6], index=['A','B','C','D','E'], name='月份')
print(s)

#通过字典来创建
s = pd.Series({'a':1,'b':2,'c':3,'d':4,'e':5},name= "月份")
print(s)
s1 = pd.Series(s, index=['a','c'])
print(s1)
```

# 2. Series 的属性

```python
print(s)
print(s.index)  #索引
print(s.values)
print(s.shape, s.ndim, s.size)
print(s.dtype,     s.name)
print('-----------')
print(s.loc['b'])   #显式索引
print(s.loc['a':'c'])
print(s.iloc[1])    #隐式索引
print(s.at['b'])
```

```python
#访问数据
print(s['a'])
print(s)
print(s[s<3])
print(s.head())     #前5
print(s.tail())     #后5
```

# 3. Series 常见函数的学习

```python
s = pd.Series([10,2,np.nan,None,3,4,5],
              index=["A","B","C","D","E","F","G"],
              name='test')
print(s)
s.head(2)
s.tail(6)
print('----------')
# 查看所有的描述性信息
print(s.describe())    #只统计有意义的数

#获取元素个数(忽略缺失值)
print(s.count() )  
#获取索引
print(s.keys())
print(s.index)

print(s.isna())     #检测缺失
print('-------------')  #检测元素是否存在
print(s.isin([4]))
```

```python
#项目1
import pandas as pd
import numpy as np
np.random.seed(20)
values = np.random.randint(50,101,10)
indexes = []
for i in range(1,11):
    indexes.append('学生'+str(i))
scores = pd.Series(values,indexes)
print(scores)
print("平均分：", scores.mean())
print("最高分：", scores.max())
print("最低分：", scores.min())
#高于平均分的学生人数
mean = scores.mean()
scores[scores>mean]
print("高于平均分的学生人数：",len(scores[scores>mean]))
print("高于平均分的学生人数：",scores[scores>mean].count())
```

# 1. DataFrame 的创建

```python
import pandas as pd
import numpy as np


#通过series来构建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df = pd.DataFrame({"第一列":s1,"第二列":s2})
df = pd.DataFrame(
    {
    "id":[1,2,3,4,5],
    "name":["tom", "mike", "jack", "bob", "alice"],
    "age":[15,17,16,14,18],
    "score":[60.5,80,46,89,76]
    },index=[1,2,3,4,5],columns=["name","age","score"]
)
df
```

# 2. DataFrame 的属性

```python
print("行索引：")
print(df.index)
print("列标签：")
print(df.columns)
print("值：")
print(df.values)

print("维度：",df.ndim)
print("形状：",df.shape)
print("数据类型:")
print(df.dtypes)

#行列转置
print(df.T)
print(df.T.index)
print("----------")
```

```python
#获取元素 loc iloc at iat 
print(df)
print("---------")
#某行
print(df.loc[4])    #index
print(df.iloc[3])   #序号
#某列
print(df.loc[:,'name']) 
print(df.iloc[:,0])
```

```python
print(df)
#当个元素
print(df.at[3,"score"])
print(df.iat[2,2])
```

```python
# 获取单列数据
print(df)
print(df["name"])
print(df.name)
print(df[["name"]]) #还是dataframe
```

```python
#查看部分数据
print(df.head(3))
print(df.tail(4))
#使用布尔索引筛选数据
df[(df.score>70) & (df.age<15)]
```

```python
#随机抽样
df.sample(3)
```

# 3. DateFrame 的常用方法

```python
print(df)
print(df.head(1))
print(df.tail(1))
print("---------")
print(df.isin(['jack',46]))
```

```python
print(df.isna())

print(df["score"].sum())
print(df.score.max())
print(df.age.min())
print(df.score.mean())      #平均数
print(df.score.median())    #中位数
print(df.score.mode())      #众数
```

```python
print(df.score.std())
print(df.score.var())
print(df.score.quantile(0.25))  #分位数
print(df.describe())
print(df.count())   #非缺失值的个数
print(df.value_counts())    #出现的次数/频率
print(df.drop_duplicates()) #去重

```

```python
print(df.duplicated(subset= "age")) #查看是否重复
df.sample() #随机抽样
print(df.replace(15,20))
```

```python
print(df)
print(df.cumsum())
```

```python
print(df.sort_index(ascending=False))
print(df.sort_values(by= ["score","age"],ascending=[False,True]))
```
