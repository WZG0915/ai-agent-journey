# 1. 折线图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams #字体
rcParams['font.family'] = 'SimHei'  #设置win默认字体
#rcParams['font.sans-serif'] = 'SimHei'

#绘制折线图

#创建图表，设置大小
plt.figure(figsize=(10,5))  #画布

#要绘制的数据
month = ['1月','2月','3月','4月']
sales = [100,150,80,130]

#绘制折线图
plt.plot(month,sales,
         label = '产品A',
         color = 'blue',
         linewidth = 1,
         linestyle = '--',
         marker = 'o'
           )


#添加标题
plt.title('2026年消费趋势', color = 'red', fontsize = 20)
#添加坐标轴的标签
plt.xlabel('月份',fontsize = 10)
plt.ylabel('销售额（万元',fontsize= 10 )
#添加图例
plt.legend(loc= 'upper left')
#添加网格线
plt.grid(True,alpha=0.1,color= 'blue',linestyle='--')         #x y均有    alpha深度
#plt.grid(axis='y')      #只有y
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=12)  #rotation控制旋转角度
plt.yticks(rotation=0,fontsize=15)
#设置y轴范围
plt.ylim(60,160)
#在每个数据点上显示数值
for x,y in zip(month,sales):
    #print(x,y)
    plt.text(x,y+2,str(y),ha='center',va='bottom',fontsize=10)    #ha 水平对齐  va 垂直对齐
#显示图表
plt.show()
```

# 2. 柱状图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams #字体
rcParams['font.family'] = 'SimHei'  #设置win默认字体
#rcParams['font.sans-serif'] = 'SimHei'



#创建图表，设置大小
plt.figure(figsize=(10,5))  #画布

#要绘制的数据
subjects = ['语文','数学','英语','科学']
scores = [85,92,78,88]

#绘制柱状图
plt.bar(subjects,scores,
        label = '小红',
        width= 0.4)
# plt.plot(month,sales,
#          label = '产品A',
#          color = 'blue',
#          linewidth = 1,
#          linestyle = '--',
#          marker = 'o'
#            )


#添加标题
plt.title('成绩分布', color = 'red', fontsize = 20)
#添加坐标轴的标签
plt.xlabel('科目',fontsize = 10)
plt.ylabel('成绩',fontsize= 10 )
#添加图例
plt.legend(loc= 'upper left')
#添加网格线
plt.grid(axis='y',alpha=0.1,color= 'blue',linestyle='--')         #x y均有    alpha深度
#plt.grid(axis='y')      #只有y
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=12)  #rotation控制旋转角度
plt.yticks(rotation=0,fontsize=15)
#设置y轴范围
plt.ylim(0,100)
#在每个数据点上显示数值
for x,y in zip(subjects,scores):
    #print(x,y)
    plt.text(x,y+2,str(y),ha='center',va='bottom',fontsize=10)    #ha 水平对齐  va 垂直对齐
#自动优化排版
plt.tight_layout()

#显示图表
plt.show()
```
